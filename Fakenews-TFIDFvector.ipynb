```python
import pandas as pd
import numpy as np
import itertools
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("news.csv",index_col=None)
df
```

```python
dataset=df.drop("Unnamed: 0",axis=1)
```

```python
dataset
```

```python
x=dataset["text"]
y=dataset["label"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.33,random_state=52)
```

```python
print(x_train)
```

```python
tfidf=TfidfVectorizer(stop_words='english', max_df=0.7) 
##Maximum Document Frequency -Any word that appears in more than 70% of your documents will be completely ignored.
tfidf_train = tfidf.fit_transform(x_train)
tfidf_test = tfidf.transform(x_test)
```

```python
len(tfidf.get_feature_names_out())  ##tells the total number of unique words (features) in your dataset
```

```python
dir(tfidf_train)
```

```python
clf=MultinomialNB()
clf.fit(tfidf_train,y_train)
pred=clf.predict(tfidf_test)
score = accuracy_score(y_test, pred)
print("accuracy:   %0.3f" % score)

cm = confusion_matrix(y_test, pred, labels=['FAKE', 'REAL'])
```

```python
report=classification_report(y_test, pred)
print(report)
```
